# Анализ симулируемой траектории
Данный ноутбук используется для построения графиков движения робота, собранных в запущенной ROS 2 симуляции. 

```python
import pandas as pd
import matplotlib.pyplot as plt

# Убедитесь, что предварительно данные были записаны с помощью record_odom.py
data_path = 'trajectory_data.csv'

try:
    df = pd.read_csv(data_path)
    
    plt.figure(figsize=(8, 6))
    plt.plot(df['x'], df['y'], 'b-', label='Simulated Odometry (with slip)', linewidth=2)
    plt.xlabel('X [m]', fontsize=12)
    plt.ylabel('Y [m]', fontsize=12)
    plt.title('Robot Trajectory Comparison', fontsize=14)
    plt.grid(True)
    plt.legend(loc='best')
    plt.axis('equal') # Для сохранения пропорций осей
    plt.tight_layout()
    
    plt.show()
except FileNotFoundError:
    print(f"Файл {data_path} не найден. Сначала запустите скрипт сбора данных.")
```
